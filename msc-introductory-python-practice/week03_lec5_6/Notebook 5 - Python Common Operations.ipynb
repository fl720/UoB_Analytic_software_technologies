#### 2. format

print() lets us display useful output. But it offered little control and you can only put a string as is. We can use .format to do more.

```python
print('A die has', 6, 'digits,',[1,2,3,4,5,6])
```

```python
s = 'A die has {} digits, {}'.format(6, [1,2,3,4,5,6])

print(s)
```

Within the {} we can place formating instructions after a :.

```python
print(1/3)

print('{:.3f}'.format(1/3))
```

```python
for i in range(8,12):
    print('{:03d}'.format(i)) # has minimum 3 digits
```

```python
for i in [1,10,-100,1000]: # try some larger numbers and negatives.
    print('{: 5d}'.format(i))
```

As well as positional arguments to format, we can also have keyword arguments. And use argument expansion to supply them.

```python
info = {'time':'morning','name':'Moody'}

print ('Good {time} {name}'.format(**info))

print ('Good {time} {name}'.format(time='morning',name='Ken')) # This is what **info looks like to format.
```

#### 3. Files

```python
# to write to a file

fout = open('log.txt', 'w') # 'w' indicates opening for writing, clear any existing contents

fout.write('Line 1\n') # we have to put in the newline
fout.write('Line {}\n'.format(2))

print('Line 3', file=fout)

fout.close()
```

```python
fin = open('log.txt', 'r') # 'r' for read

for line in fin.readlines():
    print(line, end='')

fin.close()
```

```python
with open('log.txt', 'a') as fout: # 'a' means append to the file
    fout.write('Line 4\n')

# file gets closed when we leave the block
```

There's a few other modes we can specify. Add 'b' for binary. w+, r+, a+ open for reading and writing.

##### 4.3 collections

```python
from collections import namedtuple

Name = namedtuple('Name', ['first','last'])

n = Name('Moody', 'Alam')

print(n.last)
```

```python
from collections import defaultdict

dic = defaultdict(lambda:8)

print(dic['hello']) # outputs 8 (the default value)

dic['hello'] += 4

print(dic['hello'])
```

##### 4.5 garbage collection

```python
import weakref
import gc

class Blah:
    pass

blah = Blah()

alt = weakref.ref(blah)

print(alt())
```

```python
del blah
gc.collect()
```

```python
print(alt())
```

##### 4.7 time

```python
import numpy
import time

start = time.process_time()

numpy.linalg.inv(numpy.random.rand(2048,2048))

end = time.process_time()

print(end - start, 'seconds')
```

##### 4.8 datetime

```python
import datetime

today = datetime.date.today()

print(today.isoformat())
print(type(today))
print(today.year)
```

```python
next_week = today + datetime.timedelta(days=7)

print(next_week.isoformat())
```

```python
now = datetime.datetime.now()

print(now.isoformat())

print(now.isoformat(' ')) # no T

print(now.isoformat(' ', 'seconds')) # no milliseconds
```
