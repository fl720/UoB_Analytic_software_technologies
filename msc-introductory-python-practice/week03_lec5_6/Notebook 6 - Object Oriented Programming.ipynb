#### 1. Objects, Objects Everywhere

So far in python, we have seen that we can group blocks of code into a function that can then be invoked when required. This allows us to think about the block of code with a level of abstraction.

Object oriented programming allows us to extend this abstraction to include data. We can wrap parts of our code and data into an abstraction known as an __object__ and defined in Python as a **class**.

We've already used a number of these classes. In fact, to Python, everything is an object.

```python
a = list() # create an object
a.append('some state') # a method
print(a[0]) # state
```

An object contains element of two kinds:

* **State** Data stored in the object. Variables
* **Methods** Functions associated with the object.

We can find out what state and methods are in an object using **dir()**

```python
a = list()

content = dir(a)

for item in content:
    print(item)
```

```python
print(dir(None))
```

```python
print(None.__dir__)
```

```python
with open('meh.txt ','w' ) as fin:
    print(dir(fin))
```

```python
def my_func():
    """This is my function"""
    return 6

print(dir(my_func))
```

Anything that starts with an _ is intended to be used inside the object only. You should not make use of it outside. For example **__add__()** in an integer object is the function that adds two integers together.

```python
# You should not use anything with an underscore without the object, but just this once...
print(my_func.__doc__)
```

We can check if an object is an instance of a particular type of class using **isinstance()**

```python
print ( isinstance (5 , int ))
print ( isinstance (5.5 , int ))
print ( isinstance (5.5 , float ))
```

All objects are part of a hierarchy. For example, both _int_ and _float_ are _numbers_.

```python
import numbers
print( isinstance (5 , numbers.Number ))
print( isinstance (5.5 , numbers.Number ))
s = 'not a number'
print( isinstance (s , numbers.Number ))
```

```python
print (type(None))
print (type(6) )
print (type('meow '))
```

#### 2. My First Class

Let's create a new kind of object or **class**. We do this by using **class** _name_ _of_ _class_ **:**

Inside the class block, we can define our class methods using **def**.

The function **__init__(self,...)** is the constructor for this class.

**self** is a reference to this class and allows us to access the elements of the class.

```python
class Name:
    def __init__(self , first , last ):
        self.first = first
        self.last = last


    def fullname(self):
        return self.first + ' ' + self.last
```

```python
print(dir(Name))
```

```python
# create some instances of the class
me = Name('Moody' , 'Alam')
anon = Name('John ' , 'Smith ')
```

```python
print(me.fullname()) # call the method
print(anon.fullname())
```

```python
print(me.first) # access the variables in the instance of the class
print(me.last)
```

```python
# update the variable
me.first = 'NewFirstName'

print(me.fullname())
```

It's possible to invoke methods by using the class name. It's not generally a good idea, but you will see it. The danger is you invoke the method on the wrong kind of object.

```python
print(Name.fullname(me))
```

#### 3. Privacy

Object can also be used to hide state from your fellow programmers. Some languages provide mechanisms to enforce this. Python relies on convention.

```python
class Name:
    def __init__(self , first , last ):
        self._first = first
        self._last = last

    def fullname(self):
        return self._first + ' ' + self._last

    def change(self, first, last):
        self._first = first
        self._last = last

nn = Name('Moody', 'Alam')
print(nn._first)
```

#### 4. Inheritance

Inheritance allows you to extend a class to create a new type of class that inherits the properties of its parent.

```python
class SortingDict  (dict): # we're extending the dict class
    def sortedkeys(self): # adding a new method
        for key in sorted(self.keys()):
            yield key # it's a generator
```

```python
sd = SortingDict()

sd['a'] = 0
sd['c'] = 1
sd['b'] = 2

print(list(sd.keys()))
```

```python
print(list(sd.sortedkeys()))
```

Typed languages only permit classes or their children to be passed as arguments of that type. This offers an important level of protection from errors. Python has no such restriction. Python won't check and only throws an error if you ask it to access something that's not present. If you want to be sure use instance(). See: _Duck Typing_

Inheritance can go down multiple levels:

```python
class Animal:
    def __init__(self):
        self.alive = True

    def die(self, killer=None):
        self.alive = False
        print(type(self), 'is dead')

class Mammal(Animal):
    pass

class Mouse(Mammal):
    pass

m = Mouse()
m.die()
```

```python
class Cat(Mammal): # another kind of mammal
    pass

c = Cat( )
c.die( )
```

Applications where there is a natural hierarchy of objects often make use of this. Functionality in a parent only has to be written once and will be available to all the children.

We add additional functionality and can override it if necessary in a child.

```python
class Animal:
    def __init__(self):
        self.alive = True

    def die(self, killer=None):
        self.alive = False
        print(type(self), 'is dead')

    def eat(self, target):
        raise NotImplementedError

class Mammal(Animal):
    pass

class Mouse(Mammal):
    pass

class Cat(Mammal):
    def eat(self, target):
        target.die(self)

jerry = Mouse()
tom = Cat()
```

```python
#lunch
tom.eat(jerry)
```

You can also have multiple inheritance, by listing multiple parents.

```python
class Ninja:
    def die(self, killer): # If you try to kill a ninja, it kills you.
        killer.die()

class NinjaMouse(Mammal,Ninja): # the order matters
    pass

nm = NinjaMouse()

tom.eat(nm)
```

```python
class Ninja:
    def die(self, killer): # If you try to kill a ninja, it kills you.
        killer.die()

class NinjaMouse(Mammal,Ninja): # the order matters
    def die(self, killer): # If you try to kill a ninja, it kills you.
        Ninja.die(self,killer)

nm = NinjaMouse()

tom.eat(nm)
```

```python
NinjaMouse.mro()
```

What about the constructor? If we have one, it replaces the parents class. We will need to make sure the parent constructor is called if still needed.

```python
class Ninja(Animal):
    def die(self, killer):
        killer.die( )

class NinjaMouse(Ninja,Mammal):
    def __init__(self):
        Ninja.__init__(self)
        Mammal.__init__(self)
        self.hungry = False
```

That has the problem that Animal.__init__ is called twice! Use super() it provides the super (aka parent) class.

```python
class NinjaMouse(Ninja,Mammal):
    def __init__(self):
        super().__init__()
        self.hungry = False
```

#### 5. Tricks

We make methods _static_. They don't get to see the object. We can also indicate setter/getter functionality

```python
class Algorithm:
    def __init__(self):
        self._solver = 'a'

    @staticmethod
    def version(): # no self
        return '1.0.0'

    @property
    def solver(self):
        return self._solver

    @solver.setter
    def solver(self,value):
        if value in ['a', 'b']:
            self._solver = value
        else:
            raise ValueError


a = Algorithm()

print(a.solver)
a.solver = 'b'
print(a.solver)
```

We can also have class variables. They are shared by all instances. Typically not edited. More likely to use this for version that static property.

```python
class Algorithm:
    version = '1.0.0'

a = Algorithm()

print(a.version)
```

We can also override operators such as addition, indexing, etc

```python
class Algorithm:
    def __init__(self):
        self._solver = 'a'

    def __str__(self):
        return ('Algorithm(solver = '+self._solver +')')

a = Algorithm()

print(a)
```
