### 5.1 Conventions used in every task

Please note that the provided NSS dataset for this coursework is a subset of the original NSS 2025 data, and has been processed specifically for this coursework.

**Read these carefully before you start.** Each convention has a number (Convention 1, Convention 2, ...), and the tasks refer to them by that number.

**Convention 1: Means.** Unless a task says "weighted", a mean is the simple (unweighted) mean over **all rows** that belong to the group in the task. A weighted mean uses `respondents` as weights.

**Convention 2: Student populations.** Because of the structure of the NSS questions (e.g., the same `population` value is repeated on many rows – one per question), the way the [Office for Students](https://www.officeforstudents.org.uk/) calculates and provides the NSS data, and the pre-processing done for this coursework, the NSS respondent and population numbers may not be exact. For example, for a given `cah1_subject`, the sum of the populations of its subcategories (the `cah2_subject` or `cah3_subject` populations) will not necessarily match the `cah1_subject` population. For this coursework, wherever you are asked to sum a population, you will take the **sum of the unique population values**. This gives an indirect measure of the sum of populations that is sufficient for the tasks in this coursework.

A population is therefore always calculated in two steps: first take the **unique** `population` values and add them up **within each group**, then add the groups together. Example 1 ([Section (e)](#scrollTo=76fa8f77)) shows this for one provider.

* Population of a **provider**: for each `cah1_subject` of the provider, sum the unique values; then add these over its subjects.
* Population of a **broad subject** across providers (`cah1_subject`): for each provider, sum the unique values over all of that provider's rows in the subject; then add these over the providers.
* Population of a **subject at one provider**: sum the unique values over that provider's rows in the subject.

**Convention 3: Eligible providers.** Where a task says "eligible", only providers whose population (calculated as in Convention 2 and Example 1) is at least `MIN_PROVIDER_POPULATION` (set in [Section (b)](#scrollTo=5615b2f3)) are used. Very small providers can score 0% or 100% from a handful of students.

**Convention 4: Rankings and ties.** Rank using the values **rounded to 6 decimal places**. If values are equal, the item that comes first alphabetically ranks first (question code, provider name, subject name, or the pair of question codes). Lists run from the **highest to the lowest** value unless the task says otherwise. Where a task works through the items of an earlier answer one by one (for example, "for each of the top 5 NSS questions found in Task 3"), the rows follow the order of that earlier answer, and each group is ordered as described above. Answers are stored rounded to 2 decimal places (3 for Task 18), so two answers can show the same value; their order is still decided by the values rounded to 6 decimal places.

**Convention 5: Rounding.** Do not round your intermediate results. The notebook rounds your final answers automatically (2 decimal places; 3 for Task 18).

**Convention 6: Standard deviation.** Wherever a task asks for a standard deviation, use the **sample** standard deviation (`ddof=1`). This is the default of the pandas `.std()` method; `np.std` uses `ddof=0` unless you set `ddof=1`. (For the PCA in Task 19 this choice does not change the result.)

````python
# @title Section (a): Answer Format Validation Utilities
#
# How the checks work
# * register_reference_data(nss) is called once at the end of the data-loading cell in Section (b). It takes a private
#   snapshot of the facts the checks need (question texts, providers, subjects, row counts, ...). Later changes to
#   `nss`, `employer_name` or `PLOT_BINS` in your own cells therefore cannot affect the checks.
# * add_to_answer(n, answer) first removes any answer stored earlier for Task n, then checks the new answer.
#   If the check fails, you are told what is wrong and asked whether you still want to store the answer. If you do not
#   confirm, Task n has no stored answer until you fix the problem and run the cell again.
# * The checks test the FORMAT of an answer and its CONSISTENCY with your earlier answers and with the data.
#   They do not tell you whether an answer is correct.
# * Some tasks use earlier answers (for example Task 4 uses Task 3). If you change an earlier answer afterwards,
#   the later tasks that used it are reported as "stored using an older answer" until you run them again.
import re, math, json, hashlib, html, pprint
import numpy as np
import pandas as pd
from IPython.display import display, HTML


class AnswerError(ValueError):
    """Raised when an answer does not follow the required format."""


N_TASKS = 19
_PLOT_TASKS = (5, 8)
# Earlier answers used by each task: task -> tasks whose answers it uses
_DEPENDS_ON = {4: (3,), 5: (3,), 7: (6,), 8: (6,), 9: (4,), 11: (10,), 14: (13,), 15: (14,)}
_REF = None            # snapshot of the data, set by register_reference_data()
_PROVENANCE = {}       # task n -> {k: fingerprint of the answer to Task k that was used when Task n was stored}
_UNCHECKED = set()      # tasks whose answer was stored after the student confirmed, although the check failed
_SEQUENCES = (list, tuple, np.ndarray, pd.Series, pd.Index)   # accepted list-like containers


# ---------------------------------------------------------------------------- reference data (snapshot taken once in Section (b))
def register_reference_data(df, employer=None):
    """Take a private snapshot of the facts used by the answer checks. Called at the end of the data-loading cell."""
    global _REF
    _REF = None
    if employer is None:
        employer = globals().get("employer_name")
    if not isinstance(employer, str):
        raise AnswerError("`employer_name` is not set - run the first cell of Section (b) before loading the data.")
    question = df["question"]
    ref = {
        "rows": len(df),
        "questions": frozenset(question.unique()),
        "q_rows": {q: int(c) for q, c in question.value_counts().items()},
        "q_range": {q: (float(lo), float(hi)) for q, (lo, hi) in df.groupby("question")["agree_pct"].agg(["min", "max"]).iterrows()},
        "providers": frozenset(df["provider"].unique()),
        "cah1": frozenset(df["cah1_subject"].unique()),
        "cah3": frozenset(df["cah3_subject"].unique()),
        "cah3_of_cah1": {s: frozenset(v) for s, v in df.groupby("cah1_subject")["cah3_subject"].unique().items()},
        "employer": employer,
        "employer_questions": frozenset(df.loc[df["provider"] == employer, "question"].unique()),
        # questions with results for each detailed subject from providers other than the employer (Task 15)
        "cah3_questions_others": {s: frozenset(v) for s, v in
                                  df[df["provider"] != employer].groupby("cah3_subject")["question"].unique().items()},
        "bins": tuple(float(x) for x in np.linspace(0, 100, 21)),    # frozen copy of PLOT_BINS: 0, 5, ..., 100
    }
    ref["codes"] = frozenset(q.split(":")[0] for q in ref["questions"])
    if not ref["employer_questions"]:
        raise AnswerError(f"employer_name {employer!r} does not match any provider in the data. "
                          "Correct it in the first cell of Section (b) and run the cells of Section (b) again.")
    _REF = ref
    _PROVENANCE.clear()
    _UNCHECKED.clear()


def _ref():
    if _REF is None:
        raise AnswerError("The reference data for the answer checks is missing. Run the cells of Section (b) "
                          "(they load the data) before the tasks.")
    return _REF


# ---------------------------------------------------------------------------- basic type checks
def _fail(n, msg):
    raise AnswerError(f"Task {n}: {msg}")


def _plural(k, word):
    return f"{k} {word}" if k == 1 else f"{k} {word}s"


def _short(x, size=60):
    r = " ".join(repr(x).split())      # one line, also for pandas objects
    return r if len(r) <= size else r[:size] + "..."


def _is_int(x):
    return isinstance(x, (int, np.integer)) and not isinstance(x, (bool, np.bool_))


def _is_real(x):
    return isinstance(x, (int, float, np.integer, np.floating)) and not isinstance(x, (bool, np.bool_))


def _int(n, x, where, lo=None, hi=None):
    if not _is_int(x):
        _fail(n, f"{where} must be an integer (got {type(x).__name__}: {_short(x)}). Use int(...) if needed.")
    x = int(x)
    if lo is not None and x < lo:
        _fail(n, f"{where} must be at least {lo} (got {x}).")
    if hi is not None and x > hi:
        _fail(n, f"{where} must be at most {hi} (got {x}).")
    return x


def _count(n, x, where):
    """A histogram count: an integer, or a float with an integer value (plt.hist returns floats such as 3.0)."""
    if _is_real(x) and math.isfinite(float(x)) and float(x).is_integer():
        x = int(x)
    return _int(n, x, where, 0)


def _real(n, x, where, lo=None, hi=None):
    if not _is_real(x):
        _fail(n, f"{where} must be a number (got {type(x).__name__}: {_short(x)}).")
    x = float(x)
    if not math.isfinite(x):
        _fail(n, f"{where} must be a finite number (got {x}).")
    if lo is not None and x < lo:
        _fail(n, f"{where} must be at least {lo} (got {x}).")
    if hi is not None and x > hi:
        _fail(n, f"{where} must be at most {hi} (got {x}).")
    return x


def _text(n, x, where):
    if not isinstance(x, str):
        _fail(n, f"{where} must be text (got {type(x).__name__}: {_short(x)}).")
    if x.strip() == "":
        _fail(n, f"{where} must not be empty.")
    return str(x)


def _seq(n, x, where, kind="a list"):
    """Convert a list-like value (list, tuple, numpy array, pandas Series or Index) to a list."""
    if x is None:
        _fail(n, f"{where} is None - assign your result first.")
    if isinstance(x, pd.DataFrame):
        _fail(n, f"{where} must be {kind}, not a DataFrame. Convert it, for example with df.values.tolist().")
    if not isinstance(x, _SEQUENCES) or (isinstance(x, np.ndarray) and x.ndim == 0):
        _fail(n, f"{where} must be {kind} (got {type(x).__name__}: {_short(x)}).")
    return list(x)


def _rows(n, x, nrows, ncols, what="answer"):
    """Check a list of rows with `ncols` items each (and exactly `nrows` rows unless nrows is None); return lists."""
    rows = _seq(n, x, f"the {what}", "a list of lists")
    if nrows is not None and len(rows) != nrows:
        _fail(n, f"the {what} must contain exactly {_plural(nrows, 'row')} (got {len(rows)}).")
    if nrows is None and len(rows) == 0:
        _fail(n, f"the {what} must not be empty.")
    out = []
    for i, row in enumerate(rows, start=1):
        row = _seq(n, row, f"row {i} of the {what}")
        if len(row) != ncols:
            _fail(n, f"row {i} of the {what} must have exactly {_plural(ncols, 'item')} (got {len(row)}).")
        out.append(row)
    return out


def _sorted_check(n, values, descending, what):
    """Values must be sorted; like the rankings (Convention 4), compare values rounded to 6 decimal places."""
    r = [round(v, 6) for v in values]
    for a, b in zip(r, r[1:]):
        if (descending and b > a) or (not descending and b < a):
            _fail(n, f"{what} must be sorted from {'highest to lowest' if descending else 'lowest to highest'}.")


def _check_groups(n, keys, order, per_group, label):
    """keys = first item of each row. They must follow `order`, with `per_group` consecutive rows per group."""
    expected = [g for g in order for _ in range(per_group)]
    if keys != expected:
        _fail(n, f"the rows must be grouped by the {label} in the same order as there, {_plural(per_group, 'row')} for each.")


# ---------------------------------------------------------------------------- names and earlier answers
def _question(n, q, where):
    _text(n, q, where)
    if q not in _ref()["questions"]:
        _fail(n, f"{where} {q!r} is not an NSS question in the data. Copy the text exactly as it appears in the data (including any spaces).")
    return str(q)


def _provider(n, p, where):
    _text(n, p, where)
    if p not in _ref()["providers"]:
        _fail(n, f"{where} {p!r} is not a provider in the data. Copy the name exactly as it appears in the data.")
    return str(p)


def _question_mean(n, q, v, where):
    """A mean agree_pct for question q: a number inside the range of the question's agree_pct values."""
    v = _real(n, v, where, 0, 100)
    lo, hi = _ref()["q_range"][q]
    if not lo - 0.01 <= v <= hi + 0.01:
        _fail(n, f"{where} ({v:g}) is outside the range of agree_pct for question {q[:3]} ({lo:g} to {hi:g}). "
                 "Means are percentages (for example 85.5), not fractions.")
    return v


def _prev(n, k, answers):
    """The (re-checked) stored answer to an earlier Task k that Task n uses."""
    if k not in answers:
        _fail(n, f"this task uses your answer to Task {k}. Complete (and run) Task {k} first.")
    try:
        return _VALIDATORS[k](answers[k], answers)
    except AnswerError:
        _fail(n, f"this task uses your answer to Task {k}, but the stored answer to Task {k} is not valid. Run Task {k} again first.")


def _fingerprint(x):
    return hashlib.sha1(json.dumps(x, sort_keys=True, default=str).encode()).hexdigest()


def _stale_report(answers):
    """Tasks whose stored answer was built on a different version of an earlier answer: {task: message}."""
    stale = {}
    for n in range(1, N_TASKS + 1):
        if n not in answers or n not in _PROVENANCE:
            continue
        for k, fp in _PROVENANCE[n].items():
            if k in stale:
                stale[n] = f"Task {n}: it uses your answer to Task {k}, which must be run again first; then run Task {n} again."
                break
            if k in answers and _fingerprint(answers[k]) != fp:
                stale[n] = f"Task {n} was stored using an older answer to Task {k} - run Task {n} again."
                break
    return stale


# ---------------------------------------------------------------------------- validators (one per task); each returns the cleaned answer
def _v1(a, A):
    return _int(1, a, "the answer", 1, len(_ref()["cah3"]))


def _v2(a, A):
    R = _ref()
    if not isinstance(a, (list, tuple)):
        _fail(2, f"answer_2 must be a list of two items [subject, population] (got {type(a).__name__}).")
    if len(a) > 0 and isinstance(a[0], _SEQUENCES):
        _fail(2, "answer_2 must be one flat list of two items [subject, population], not a list of lists.")
    if len(a) != 2:
        _fail(2, f"answer_2 must be a list of exactly two items [subject, population] (got {_plural(len(a), 'item')}).")
    s = _text(2, a[0], "the subject (first item)")
    if s not in R["cah3"]:
        _fail(2, f"{s!r} is not a cah3_subject in the data (copy it exactly).")
    return [s, _int(2, a[1], "the population (second item)", 1)]


def _v_question_list(n, a, nrows, descending, what="answer"):
    """Rows [question, mean]: different questions, sorted by the mean. Used by Tasks 3, 6, 15 and 16."""
    rows = _rows(n, a, nrows, 2, what)
    out, seen = [], set()
    for i, (q, v) in enumerate(rows, start=1):
        q = _question(n, q, f"row {i} question")
        if q in seen:
            _fail(n, f"question {q!r} appears more than once.")
        seen.add(q)
        out.append([q, _question_mean(n, q, v, f"row {i} mean")])
    _sorted_check(n, [v for _, v in out], descending, "the rows")
    return out


def _v3(a, A):
    return [[q, round(v, 2)] for q, v in _v_question_list(3, a, 5, True)]


def _v6(a, A):
    return [[q, round(v, 2)] for q, v in _v_question_list(6, a, 5, False)]


def _v_provider_rankings(n, a, A, src, descending):
    order = [r[0] for r in _prev(n, src, A)]
    rows = _rows(n, a, 15, 3)
    questions = [_question(n, r[0], f"row {i} question") for i, r in enumerate(rows, start=1)]
    _check_groups(n, questions, order, 3, f"questions of Task {src}")
    out, seen = [], set()
    for i, (q, (_, p, v)) in enumerate(zip(questions, rows), start=1):
        p = _provider(n, p, f"row {i} provider")
        if (q, p) in seen:
            _fail(n, f"provider {p!r} appears twice for the same question.")
        seen.add((q, p))
        out.append([q, p, _question_mean(n, q, v, f"row {i} mean")])
    for k in range(5):
        _sorted_check(n, [r[2] for r in out[3 * k:3 * k + 3]], descending, f"the 3 providers of question {out[3 * k][0][:3]}")
    return [[q, p, round(v, 2)] for q, p, v in out]


def _v4(a, A):
    return _v_provider_rankings(4, a, A, 3, True)


def _v7(a, A):
    return _v_provider_rankings(7, a, A, 6, False)


def _v_distribution(n, a, A, src):
    """Rows [question, mean, sd] for the 5 questions of Task src; each mean must agree with the Task src mean."""
    earlier = _prev(n, src, A)
    rows = _rows(n, a, 5, 3)
    questions = [_question(n, r[0], f"row {i} question") for i, r in enumerate(rows, start=1)]
    if questions != [r[0] for r in earlier]:
        _fail(n, f"the questions must be the 5 questions of Task {src}, in the same order.")
    out = []
    for i, (q, (_, m, s), (_, m_src)) in enumerate(zip(questions, rows, earlier), start=1):
        m = round(_real(n, m, f"row {i} mean", 0, 100), 2)
        s = round(_real(n, s, f"row {i} standard deviation", 0, 50), 2)
        if s <= 0:
            _fail(n, f"row {i} standard deviation must be greater than 0 (got {s}).")
        if abs(m - m_src) > 0.011:
            _fail(n, f"row {i}: the mean {m:.2f} differs from your Task {src} mean {m_src:.2f} for the same question. "
                     f"Both are the simple mean over all rows of the question, so they must agree.")
        out.append([q, m, s])
    return out


def _v_plot(n, plot, answer_rows):
    """Histogram counts of Tasks 5 and 8, checked against the row counts and against the mean and sd of the answer."""
    R = _ref()
    edges = R["bins"]
    nbins = len(edges) - 1
    rows = _rows(n, plot, 5, 2, "plot data")
    questions = [_question(n, r[0], f"plot row {i} question") for i, r in enumerate(rows, start=1)]
    if questions != [r[0] for r in answer_rows]:
        _fail(n, f"the plot data must list the same 5 questions in the same order as answer_{n}.")
    out = []
    for i, (q, (_, counts), (_, mean, sd)) in enumerate(zip(questions, rows, answer_rows), start=1):
        counts = _seq(n, counts, f"plot row {i}: the counts (a list of {nbins} integers)")
        if len(counts) != nbins:
            _fail(n, f"plot row {i}: expected {nbins} histogram counts (use bins=PLOT_BINS, the 21 edges 0, 5, ..., 100), got {len(counts)}.")
        counts = [_count(n, c, f"plot row {i} count {j}") for j, c in enumerate(counts, start=1)]
        total = R["q_rows"][q]
        if sum(counts) != total:
            _fail(n, f"plot row {i}: the counts add up to {sum(counts)} but the question has {total} rows. "
                     "The histogram must cover every row of the question (check bins=PLOT_BINS).")
        low = sum(c * edges[j] for j, c in enumerate(counts)) / total
        high = sum(c * edges[j + 1] for j, c in enumerate(counts)) / total
        if not low - 0.011 <= mean <= high + 0.011:
            _fail(n, f"plot row {i}: these counts put the mean of the question between {low:.2f} and {high:.2f}, "
                     f"but your mean in answer_{n} is {mean:.2f}. The counts must be in bin order (lowest bin first) "
                     "and come from all rows of the same question.")
        used = [j for j, c in enumerate(counts) if c > 0]
        spread = edges[used[-1] + 1] - edges[used[0]]
        sd_max = spread / 2 * math.sqrt(total / (total - 1)) if total > 1 else 0.0
        if sd > sd_max + 0.011:
            _fail(n, f"plot row {i}: these counts only cover agree_pct from {edges[used[0]]:g} to {edges[used[-1] + 1]:g}, "
                     f"which allows a standard deviation of at most {sd_max:.2f}, but your standard deviation in answer_{n} is {sd:.2f}.")
        out.append([q, counts])
    return out


def _v5(a, A):
    return _v_distribution(5, a, A, 3)


def _v8(a, A):
    return _v_distribution(8, a, A, 6)


def _v9(a, A):
    # Only the link with Task 4 is checked (the same providers, each once). Populations are not compared with
    # MIN_PROVIDER_POPULATION here: eligibility is part of Task 4, not of Task 9.
    expected = {r[1] for r in _prev(9, 4, A)}
    rows = _rows(9, a, None, 2)
    out, seen = [], set()
    for i, (p, pop) in enumerate(rows, start=1):
        p = _provider(9, p, f"row {i} provider")
        if p in seen:
            _fail(9, f"provider {p!r} appears more than once - list each provider only once.")
        seen.add(p)
        out.append([p, _int(9, pop, f"row {i} population", 1)])
    if seen != expected:
        _fail(9, f"the providers must be exactly the providers found in Task 4, each listed once "
                 f"({_plural(len(expected), 'provider')}; you gave {len(seen)}).")
    keys = [(-r[1], r[0]) for r in out]
    if keys != sorted(keys):
        _fail(9, "sort by population (largest first); providers with equal populations by name A-Z.")
    return out


def _v10(a, A):
    R = _ref()
    rows = _rows(10, a, 5, 2)
    out, seen = [], set()
    for i, (s, pop) in enumerate(rows, start=1):
        s = _text(10, s, f"row {i} subject")
        if s not in R["cah1"]:
            _fail(10, f"{s!r} is not a cah1_subject in the data.")
        if s in seen:
            _fail(10, f"subject {s!r} appears more than once.")
        seen.add(s)
        out.append([s, _int(10, pop, f"row {i} population", 1)])
    _sorted_check(10, [r[1] for r in out], True, "the rows")
    return out


def _v11(a, A):
    order = [r[0] for r in _prev(11, 10, A)]
    rows = _rows(11, a, 15, 3)
    subjects = [_text(11, r[0], f"row {i} subject") for i, r in enumerate(rows, start=1)]
    _check_groups(11, subjects, order, 3, "subjects of Task 10")
    out, seen = [], set()
    for i, (s, (_, q, v)) in enumerate(zip(subjects, rows), start=1):
        q = _question(11, q, f"row {i} question")
        if (s, q) in seen:
            _fail(11, f"question {q[:3]} appears twice for subject {s!r}.")
        seen.add((s, q))
        out.append([s, q, _question_mean(11, q, v, f"row {i} mean")])
    for k in range(5):
        _sorted_check(11, [r[2] for r in out[3 * k:3 * k + 3]], True, f"the 3 questions of subject {out[3 * k][0]!r}")
    return [[s, q, round(v, 2)] for s, q, v in out]


def _v12(a, A):
    R = _ref()
    rows = _rows(12, a, 5, 2)
    out, seen = [], set()
    for i, (s, v) in enumerate(rows, start=1):
        s = _text(12, s, f"row {i} subject")
        if s not in R["cah3"]:
            _fail(12, f"{s!r} is not a cah3_subject in the data.")
        if s in seen:
            _fail(12, f"subject {s!r} appears more than once.")
        seen.add(s)
        out.append([s, _real(12, v, f"row {i} mean", 0, 100)])
    _sorted_check(12, [r[1] for r in out], True, "the rows")
    return [[s, round(v, 2)] for s, v in out]


def _v13(a, A):
    s = _text(13, a, "the answer")
    if s not in _ref()["cah1"]:
        _fail(13, f"{s!r} is not a cah1_subject in the data (copy it exactly).")
    return s


def _v14(a, A):
    R = _ref()
    s = _text(14, a, "the answer")
    if s not in R["cah3"]:
        _fail(14, f"{s!r} is not a cah3_subject in the data (copy it exactly).")
    broad = _prev(14, 13, A)
    if s not in R["cah3_of_cah1"].get(broad, frozenset()):
        _fail(14, f"{s!r} is not a subject within your Task 13 broad subject {broad!r}.")
    return s


def _v15(a, A):
    subject = _prev(15, 14, A)
    out = _v_question_list(15, a, 5, True)
    available = _ref()["cah3_questions_others"].get(subject, frozenset())
    for q, _ in out:
        if q not in available:
            _fail(15, f"question {q[:3]} has no results for your chosen subject {subject!r} (outside your employer).")
    return [[q, round(v, 2)] for q, v in out]


def _v16(a, A):
    R = _ref()
    expected = R["employer_questions"]
    out = _v_question_list(16, a, len(expected), True, "answer (one row for each question your employer has results for)")
    if {q for q, _ in out} != expected:
        _fail(16, f"the answer must contain every question your employer has results for ({_plural(len(expected), 'question')}).")
    return [[q, round(v, 2)] for q, v in out]


def _v17(a, A):
    rows = _rows(17, a, 10, 2)
    out, seen = [], set()
    for i, (p, g) in enumerate(rows, start=1):
        p = _provider(17, p, f"row {i} provider")
        if p in seen:
            _fail(17, f"provider {p!r} appears more than once.")
        seen.add(p)
        out.append([p, _real(17, g, f"row {i} gap", -100, 100)])
    top, low = [r[1] for r in out[:5]], [r[1] for r in out[5:]]
    _sorted_check(17, top, True, "rows 1-5 (largest gaps)")
    _sorted_check(17, low, False, "rows 6-10 (smallest gaps)")
    if round(max(low), 6) > round(min(top), 6):
        _fail(17, "the 5 'smallest' gaps must not exceed the 5 'largest' gaps.")
    return [[p, round(g, 2)] for p, g in out]


def _v18(a, A):
    R = _ref()
    rows = _rows(18, a, 5, 3)
    out, seen = [], set()
    for i, (a1, a2, r) in enumerate(rows, start=1):
        for code in (a1, a2):
            _text(18, code, f"row {i} question code")
            if not re.fullmatch(r"Q[0-9]{2}", code):
                _fail(18, f"row {i}: {code!r} is not a question code like 'Q05' (use the text before the colon, nothing else).")
            if code not in R["codes"]:
                _fail(18, f"row {i}: {code!r} is not a question code in the data.")
        if not a1 < a2:
            _fail(18, f"row {i}: the first code must be the lower-numbered question ({a1!r} vs {a2!r}).")
        if (a1, a2) in seen:
            _fail(18, f"the pair {a1}-{a2} appears more than once.")
        seen.add((a1, a2))
        r = _real(18, r, f"row {i} correlation", -1 - 1e-9, 1 + 1e-9)     # allow floating-point noise such as 1.0000000000000002
        out.append([str(a1), str(a2), min(1.0, max(-1.0, r))])
    _sorted_check(18, [r[2] for r in out], True, "the rows")
    return [[x, y, round(r, 3)] for x, y, r in out]


def _v19(a, A):
    R = _ref()
    p = len(R["codes"])          # PCA on the table has at most this many components
    tol = 0.02                   # allows for the rounding of the stored percentages
    if not isinstance(a, dict):
        _fail(19, f"the answer must be a dictionary (got {type(a).__name__}).")
    keys = {"variance_explained_pct", "n_components_80pct", "top5_providers"}
    if set(a) != keys:
        _fail(19, f"the dictionary must have exactly the keys {sorted(keys)} "
                  f"(missing: {sorted(keys - set(a), key=str)}, unexpected: {sorted(set(a) - keys, key=str)}).")
    v = _seq(19, a["variance_explained_pct"], "variance_explained_pct")
    if len(v) != 3:
        _fail(19, f"variance_explained_pct must be a list of 3 numbers (PC1, PC2, PC3), got {_plural(len(v), 'item')}.")
    v = [_real(19, x, f"variance_explained_pct[{i}]", 0, 100) for i, x in enumerate(v)]
    if min(round(x, 2) for x in v) <= 0:
        _fail(19, "variance_explained_pct values must be greater than 0 after rounding to 2 decimal places (they are percentages, e.g. 12.34).")
    _sorted_check(19, v, True, "variance_explained_pct (PC1 >= PC2 >= PC3)")
    if sum(v) > 100 + 1e-6:
        _fail(19, "the three percentages cannot add up to more than 100.")
    if sum(v) <= 1:
        _fail(19, "variance_explained_pct looks like ratios (fractions of 1). Give percentages: multiply the ratios by 100.")
    if v[0] < 100 / p - tol:
        _fail(19, f"PC1 must explain at least 100/{p} = {100 / p:.2f}% of the variance, because it is the largest of at most {p} components "
                  f"(got {v[0]:.2f}). Give percentages: multiply the ratios by 100.")
    n80 = _int(19, a["n_components_80pct"], "n_components_80pct", 1, p)
    cum = [v[0], v[0] + v[1], v[0] + v[1] + v[2]]
    reached = [k + 1 for k in range(3) if cum[k] >= 80 + tol]
    missed = [k + 1 for k in range(3) if cum[k] < 80 - tol]
    if reached and n80 > reached[0]:
        _fail(19, f"PC1 to PC{reached[0]} already explain {cum[reached[0] - 1]:.2f}% (at least 80%), "
                  f"so n_components_80pct must be at most {reached[0]} (got {n80}).")
    if missed and n80 <= missed[-1]:
        _fail(19, f"PC1 to PC{missed[-1]} explain only {cum[missed[-1] - 1]:.2f}% (less than 80%), "
                  f"so n_components_80pct must be greater than {missed[-1]} (got {n80}).")
    t5 = _seq(19, a["top5_providers"], "top5_providers")
    if len(t5) != 5:
        _fail(19, f"top5_providers must be a list of 5 provider names (got {_plural(len(t5), 'item')}).")
    t5 = [_provider(19, x, f"top5_providers[{i}]") for i, x in enumerate(t5)]
    if len(set(t5)) != 5:
        _fail(19, "top5_providers must contain 5 different providers.")
    return {"variance_explained_pct": [round(x, 2) for x in v], "n_components_80pct": n80, "top5_providers": t5}


_VALIDATORS = {1: _v1, 2: _v2, 3: _v3, 4: _v4, 5: _v5, 6: _v6, 7: _v7, 8: _v8, 9: _v9, 10: _v10,
               11: _v11, 12: _v12, 13: _v13, 14: _v14, 15: _v15, 16: _v16, 17: _v17, 18: _v18, 19: _v19}


# ---------------------------------------------------------------------------- public functions used in the notebook
def _box(colour, background, text_colour, content):
    return (f'<div style="border-left:5px solid {colour};background:{background};color:{text_colour};'
            f'padding:8px 14px;border-radius:4px;margin:6px 0;">{content}</div>')


def _jsonable(x):
    """Convert an answer (possibly holding numpy or pandas objects) to plain Python values that can be saved as JSON."""
    if isinstance(x, (pd.Series, pd.Index)):
        x = x.tolist()
    elif isinstance(x, pd.DataFrame):
        x = x.values.tolist()
    elif isinstance(x, np.ndarray):
        x = x.tolist()
    if isinstance(x, dict):
        return {str(k): _jsonable(v) for k, v in x.items()}
    if isinstance(x, (list, tuple, set, frozenset)):
        return [_jsonable(v) for v in x]
    if isinstance(x, (bool, np.bool_)):
        return bool(x)
    if isinstance(x, (int, np.integer)):
        return int(x)
    if isinstance(x, (float, np.floating)):
        return float(x) if math.isfinite(float(x)) else None
    if x is None or isinstance(x, str):
        return x
    return str(x)


def _describe(a):
    """A short description of the structure of an answer, for example 'a list of 5 lists'."""
    if isinstance(a, bool):
        return "a Boolean"
    if isinstance(a, int):
        return "an integer"
    if isinstance(a, float):
        return "a number"
    if isinstance(a, str):
        return "a string"
    if isinstance(a, dict):
        return "a dictionary with the keys " + ", ".join(repr(k) for k in a)
    if isinstance(a, list):
        kinds = {type(x).__name__ for x in a}
        if len(kinds) == 1 and kinds != {"list"} or not a:
            return f"a list of {_plural(len(a), 'item')}"
        if kinds == {"list"}:
            return f"a list of {_plural(len(a), 'list')}"
        return f"a list of {_plural(len(a), 'item')}"
    return f"a {type(a).__name__}"


def _offer_to_store(n, answer, plot_data, problem):
    """The check failed: show the problem and let the student decide whether to store the answer anyway.
    Raises `problem` if the answer is not stored, so that an automatic run of the notebook still stops."""
    display(HTML(_box("#B91C1C", "#FDE8E8", "#4A0E0E",
                      f"<b>Task {n}: the answer does not pass the format check.</b><br>{html.escape(str(problem))}")))
    print()
    print(f"Task {n}: do you still want to store this answer?")
    print("  - Type yes and press Enter to store it as it is.")
    print(f"  - Just press Enter to leave Task {n} without an answer.")
    try:
        reply = input("Your choice: ")
    except Exception:                      # no keyboard input is possible (for example when the notebook runs automatically)
        raise problem from None
    if reply.strip().lower() not in ("yes", "y"):
        print(f"Task {n}: nothing was stored. Correct the answer and run the cell again.")
        return False
    all_answers[n] = _jsonable(answer)
    if n in _PLOT_TASKS:
        all_answers[f"plot_{n}"] = _jsonable(plot_data)
    _UNCHECKED.add(n)
    _PROVENANCE[n] = {k: _fingerprint(all_answers[k]) for k in _DEPENDS_ON.get(n, ()) if k in all_answers}
    display(HTML(_box("#B45309", "#FEF3C7", "#3B2A05",
                      f"<b>Task {n}: your answer was stored as it is, but it did not pass the format check.</b> "
                      "It may be marked as incorrectly formatted. To fix it, change your answer and run this cell again.")))
    return True


def add_to_answer(answer_number, answer, plot_data=None):
    """Check an answer, round it and store it in `all_answers`. Tasks 5 and 8 also need `plot_data`."""
    n = answer_number
    if not _is_int(n) or not 1 <= int(n) <= N_TASKS:
        raise AnswerError(f"The task number must be an integer from 1 to {N_TASKS} (got {n!r}).")
    n = int(n)
    if "all_answers" not in globals():
        raise AnswerError("`all_answers` is not defined - run the cells of Section (b) first.")
    # Forget any earlier answer to this task first, so that a failed re-run cannot leave an old answer behind
    all_answers.pop(n, None)
    all_answers.pop(f"plot_{n}", None)
    _PROVENANCE.pop(n, None)
    _UNCHECKED.discard(n)
    _ref()
    if answer is None:
        _fail(n, f"your answer is still None - assign your result to answer_{n} first.")
    if n in _PLOT_TASKS and plot_data is None:
        _fail(n, f"also pass your histogram counts: add_to_answer({n}, answer_{n}, plot_{n}).")
    if n not in _PLOT_TASKS and plot_data is not None:
        _fail(n, "this task has no plot data.")
    try:
        clean = _VALIDATORS[n](answer, all_answers)
        clean_plot = _v_plot(n, plot_data, clean) if n in _PLOT_TASKS else None
    except AnswerError as e:
        _offer_to_store(n, answer, plot_data, e)
        return
    except Exception as e:          # safety net: report any unexpected problem as an answer error
        _offer_to_store(n, answer, plot_data, AnswerError(
            f"Task {n}: the answer could not be checked ({type(e).__name__}: {e}). "
            "Check that it has exactly the format given in the task."))
        return
    all_answers[n] = clean
    if clean_plot is not None:
        all_answers[f"plot_{n}"] = clean_plot
    _PROVENANCE[n] = {k: _fingerprint(all_answers[k]) for k in _DEPENDS_ON.get(n, ())}
    stale = _stale_report(all_answers)
    if stale:
        display(HTML(_box("#B45309", "#FEF3C7", "#3B2A05", "<b>Note: some later answers used an older version of an answer.</b><ul style=\"margin:4px 0 0 0;\">"
                          + "".join(f"<li>{html.escape(m)}</li>" for m in stale.values()) + "</ul>")))


def show_answer(n):
    """Display the answer that has been stored for task n, in the format in which it is stored."""
    if n not in all_answers:
        print(f"No answer is stored for Task {n}.")
        return
    a = all_answers[n]
    if n in _UNCHECKED:
        display(HTML(_box("#B45309", "#FEF3C7", "#3B2A05",
                          f"<b>Task {n}: Your answer is stored as</b> {html.escape(_describe(a))} (it did not pass the format check):")))
    else:
        display(HTML(_box("#0F766E", "#E3F4F1", "#0B3B36",
                          f"<b>Task {n}: Your answer is stored as</b> {html.escape(_describe(a))}:")))
    print(pprint.pformat(a, width=110, sort_dicts=False))


def validate_all_answers(answers):
    """Check every answer again from scratch. Returns a list of problems (an empty list means everything is fine)."""
    if _REF is None:
        return ["The reference data for the answer checks is missing. Run the cells of Section (b) and then every task again."]
    if not isinstance(answers, dict):
        return [f"The answers must be a dictionary (got {type(answers).__name__})."]
    problems, norm, unexpected = [], {}, []
    for k, v in answers.items():
        if isinstance(k, (bool, np.bool_)):
            unexpected.append(k)
            continue
        key = int(k) if _is_int(k) or (isinstance(k, str) and k.isascii() and k.isdigit()) else k
        if key in norm:
            problems.append(f"Task {key}: the answers contain this task twice (keys {key!r} and {k!r}).")
        norm[key] = v
    for n in range(1, N_TASKS + 1):
        if n not in norm:
            problems.append(f"Task {n}: no answer stored (run the cell of Task {n}).")
            continue
        try:
            clean = _VALIDATORS[n](norm[n], norm)
            if n in _PLOT_TASKS:
                if f"plot_{n}" not in norm:
                    problems.append(f"Task {n}: histogram data (plot_{n}) is missing.")
                else:
                    _v_plot(n, norm[f"plot_{n}"], clean)
        except AnswerError as e:
            problems.append(str(e))
        except Exception as e:      # a malformed answer must be reported, never crash the check
            problems.append(f"Task {n}: could not be checked ({type(e).__name__}: {e}). Run Task {n} again.")
    problems += list(_stale_report(norm).values())
    unexpected += [k for k in norm if not (isinstance(k, int) and 1 <= k <= N_TASKS) and k not in ("plot_5", "plot_8")]
    if unexpected:
        problems.append(f"Unexpected entries in the answers: {unexpected}")
    return problems
````

## Section (e): Worked example (Example 1): Student population at your employer

For the purpose of this coursework, this example demonstrates how to calculate the number of students correctly when the dataset is grouped by a particular column (or criteria) resulting in multiple population values (i.e., a list of populations). The total population must be calculated as the **sum of unique values within the list of populations**.

This example shows how to sum the unique populations when grouped by broad subject (`cah1_subject`).

````python
# Example 1: total population of your employer, using unique population values within each broad subject

# Filtering for the employer
employer_rows = nss[nss["provider"] == employer_name]

# Group by 'cah1_subject', summing unique population values, and sorting by sum of unique population
by_subject = (employer_rows
              .groupby("cah1_subject", as_index=False)
              # We create a new column "sum_unique_population" which uses a lambda function to sum the unique values (via set() function)
              .agg(sum_unique_population=("population", lambda x: sum(set(x))))
              .sort_values(by="sum_unique_population", ascending=False))

display(by_subject.head())
print("Number of broad subjects =", by_subject.shape[0])
print("Total population of", employer_name, "=", by_subject["sum_unique_population"].sum())
````
