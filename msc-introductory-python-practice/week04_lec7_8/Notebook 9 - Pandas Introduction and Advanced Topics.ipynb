## 1. Why Pandas?

Data rarely arrives in a form ready for analysis.  
In research, business, and engineering, data is typically **tabular** — rows and columns — but also **messy**, **incomplete**, and often **massive**.  
Pandas was created to solve this problem efficiently, combining the power of **NumPy** (for fast numerical computation) with the convenience of **labels** and **metadata**, much like a spreadsheet or a database table.

### What Pandas brings to Python
- **Speed** — built on NumPy arrays for efficient computation.  
- **Expressiveness** — concise, readable syntax that abstracts away loops.  
- **Alignment & Labels** — operations respect row/column names, reducing logical errors.  
- **Integration** — works seamlessly with NumPy, Matplotlib, scikit-learn, and SQL.  
- **Flexibility** — handles numerical, categorical, text, datetime, and mixed-type data.

Pandas essentially gives Python the capabilities of R’s `data.frame`, Excel’s tables, and SQL’s joins — in one unified framework.

---
> **In essence, pandas is the *Swiss Army knife* of tabular data manipulation in Python —  the bridge between raw data and insight.**

**When pandas is *not* enough.** pandas holds the whole table in RAM, and a rule of thumb is that you need roughly **5-10× the file size** in memory once you start transforming it. Section 14 shows what to do when your data outgrows one machine.

### `pandas` Common Use Cases

| Stage | Task | pandas Feature |
|--------|------|----------------|
| **Data ingestion** | Read CSV, Excel, Parquet, or SQL tables | `read_csv()`, `read_excel()`, `read_parquet()`, `read_sql()` |
| **Data cleaning** | Handle missing values, fix dtypes, standardise labels | `.fillna()`, `.dropna()`, `.astype()`, `.rename()` |
| **Transformation** | Create or modify columns | `.assign()`, `.apply()`, `.map()`, `.replace()` |
| **Aggregation & summarisation** | Group and compute summaries | `.groupby()`, `.agg()`, `.pivot_table()` |
| **Reshaping** | Convert between wide and long formats | `.melt()`, `.pivot()` |
| **Integration** | Combine multiple data sources | `.merge()`, `.concat()`, `.join()` |
| **Export** | Save results to file formats for others | `.to_csv()`, `.to_parquet()`, `.to_excel()` |

## 4. pandas Fundamentals: DataFrame & Series

At the heart of pandas are two core data structures — **Series** and **DataFrame** — both designed to make tabular data manipulation intuitive and efficient.

---

### **Series**
A **Series** is a **one-dimensional labelled array** that can hold any data type — integers, floats, strings, booleans, or even Python objects.  
You can think of it as a single column in a spreadsheet, with both **values** and **labels** (the index).

In the example below:
- The **name** is the name of the series.
- The **values** are the data (`"Alice"`, `"Bob"`, `"Charlie"`).  
- The **index** (`0, 1, 2`) provides a label for each value.

````python
# Create a pandas Series
s = pd.Series(["Alice", "Bob", "Charlie"], name="passenger")
display(s)
````

````python
s1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'], name='s1')
display(s1)
s2 = pd.Series([10, 20, 30], index=['b', 'c', 'd'], name='s2')
display(s2)
print("\n")
print("s1 + s2:")
display(s1 + s2)
````

**What just happened?** `s1` has labels `a, b, c` and `s2` has `b, c, d`. pandas **aligns on the labels**, not on the positions, so only `b` (2 + 10) and `c` (3 + 20) have a partner. `a` and `d` have no partner in the other Series, so the result is `NaN` for them. If you would rather treat a missing partner as 0, use the method form with `fill_value`.

````python
# the method form of '+' lets us decide what a missing partner means
display(s1.add(s2, fill_value=0))
````



### **DataFrame**
A **DataFrame** is a **two-dimensional table** made up of multiple Series that share a common index.  
It’s conceptually similar to an Excel sheet or an SQL table — rows represent **records**, and columns represent **variables**.

Each column in a DataFrame is itself a Series object.  
This makes it easy to manipulate columns individually while maintaining the context of the full dataset.

Unlike NumPy arrays, which are purely positional, pandas objects use **labels** (row and column names) to align data automatically.  
This alignment ensures that operations between objects — like addition, subtraction, or joins — happen by label rather than by position.



````python
# A single column is a Series
ages = titanic['age']
print(type(ages), ages.shape)
display(ages.head())

# A DataFrame is 2D
print(type(titanic), titanic.shape)
titanic.head(3)
````

**Building a DataFrame by hand.** A dictionary of lists is the quickest way: each key becomes a column. Note again the difference between **one pair of brackets** (a Series) and **two pairs** (a DataFrame).

````python
mini = pd.DataFrame(
    {"name": ["Alice", "Bob", "Charlie"], "age": [29, 41, 35], "survived": [1, 0, 1]},
    index=["p1", "p2", "p3"],
)
display(mini)
print(mini.dtypes)
print(type(mini["age"]), type(mini[["age"]]))   # [] -> Series ; [[]] -> DataFrame
````

## 6. Selecting, Indexing, and Filtering
There are three primary ways to select data:
- `[]` for simple column selection
- `.loc[row_selector, col_selector]` for **label-based** selection
- `.iloc[row_idx, col_idx]` for **position-based** selection

Use **Boolean masks** to filter rows, and prefer `.loc` when setting values.

````python
# Column selection
subset = titanic[['age', 'fare']]
display(subset.head())
````

````python
# Label-based selection with one condition and one column
subset = titanic.loc[(titanic['age'] > 30), 'fare']
print("Label-based single-column subset:", subset.shape)
display(subset.head())
````

````python
# Label-based selection with multiple conditions
subset = titanic.loc[
    (titanic['age'] > 30) & (titanic['sex'] == 'female'),
    ['age', 'sex', 'class', 'survived']
]
print("Label-based selection with multiple conditions and multi-column subset:", subset.shape)
display(subset.head())
````

````python
# Index-based (position) selection using .iloc
# Syntax: df.iloc[row_index, column_index]
# Example: first 5 rows, first 3 columns (by position)
subset_iloc = titanic.iloc[0:5, 0:3]
print("Index-based subset (iloc):", subset_iloc.shape)
display(subset_iloc)
````

````python
# Another example: every 10th row, columns 1–4
subset_iloc2 = titanic.iloc[::10, 1:5]
print("Sampled index-based subset:", subset_iloc2.shape)
display(subset_iloc2.head())
````

### `.loc` vs `.iloc`: why the difference matters

So far the index is the default `0, 1, 2, ...`, so labels and positions are **identical** and `.loc` and `.iloc` look the same. They differ as soon as the index is not 0..n-1, for example after filtering. Also note that a **label** slice with `.loc` *includes* the end label, whereas a **position** slice with `.iloc` *excludes* the end position.

````python
older = titanic[titanic["age"] > 60]            # filtered frame: the index labels are NOT 0, 1, 2, ...
display(older[["age", "fare"]].head(3))

first_label = older.index[0]
print("first label                :", first_label)
print(".loc[first_label]          :", older.loc[first_label, "age"])     # label-based
print(".iloc[0]                   :", older.iloc[0]["age"])              # position-based: the same row here
print("is label 0 in this index?  :", 0 in older.index)                  # so older.loc[0] would raise a KeyError

# slicing: loc is inclusive of the end label, iloc is exclusive of the end position
print(len(titanic.loc[10:12]), "rows with loc[10:12]  vs ", len(titanic.iloc[10:12]), "rows with iloc[10:12]")
````

### Filtering helpers: `isin()`, `between()` and `query()`

Boolean masks with `&` and `|` become long and hard to read. pandas offers helpers for common filters:

- `s.isin([...])` tests membership in a list (a readable replacement for many `|` conditions).
- `s.between(a, b)` tests `a <= s <= b`.
- `df.query("age > 30 and sex == 'female'")` writes a whole condition as a string (inside the string use `and`/`or`; refer to Python variables with `@`). Try the first two helpers before this one.

**Common pitfall:** `&` and `|` bind more tightly than `>` or `==`, so every condition in a mask needs its own parentheses.

````python
print("isin            :", titanic[titanic["embark_town"].isin(["Cherbourg", "Queenstown"])].shape)
print("between (20-30) :", titanic[titanic["age"].between(20, 30)].shape)

q = titanic.query("age > 30 and sex == 'female'")               # same rows as the .loc example above
print("query           :", q.shape)

# the precedence pitfall: without parentheses Python evaluates  30 & titanic['sex']  first
try:
    titanic[titanic["age"] > 30 & titanic["sex"] == "female"]
except Exception as e:
    print("Error without parentheses:", type(e).__name__)
````

#### **Chained assignment issues**

A Boolean mask usually produces a new copy of the filtered data, not a view of the original DataFrame.

When you write
```python
subset = titanic[titanic["age"] > 60]
```

`subset` is a new DataFrame that holds a copy of the rows where `age > 60`.
It’s not a live view into titanic — changing subset will not automatically update titanic.

If you try to set values on filtered data, pandas issues a warning.

```python
titanic[titanic["age"] > 60]["fare"] = 99 # will generate a warning
```

Here, `titanic[titanic["age"] > 60]` produces a temporary filtered copy,
and then `["fare"] = 99` tries to modify it.
But since pandas isn’t sure if it’s a view or a copy, it warns you.


The correct, unambiguous way:
```python
titanic.loc[titanic["age"] > 60, "fare"] = 99
```

Here, `.loc` directly references the original DataFrame and performs the update safely


In summary,

| Operation | Returns | Modifies Original? |
|------------|----------|--------------------|
| `df[df["age"] > 60]` | Usually a **copy** |  No |
| `df.loc[df["age"] > 60, "fare"] = 99` | Works on the **original DataFrame** |  Yes |
| `df.iloc[0:5]` | May return a **view or copy** depending on internal layout |  Not guaranteed |

---

**Key takeaways:**
- Boolean masks **typically create a copy**, not a view of the original dataset.  
- Modifying data through a Boolean mask (e.g., `df[df["col"] > x]["other_col"] = ...`) may not affect the source DataFrame.  
- Always use **`.loc`** for safe, unambiguous selection and assignment — it ensures you’re working directly on the original DataFrame.

````python
titanic[titanic["age"] > 60]["fare"] = 99 # will generate a warning
````

**Check what happened.** The cell above printed a warning (pandas 2: `SettingWithCopyWarning`; pandas 3: `ChainedAssignmentError`) and, as the warning says, it did **not** change `titanic`. Under **Copy-on-Write** (the default in pandas 3) the rule is simple: *every* result of an indexing operation behaves like a copy, so `df[mask]["col"] = value` can never update `df`. The table above says `.iloc[0:5]` "may return a view or a copy": that was the pandas 2 situation; with pandas 3 it always behaves as a copy.

````python
# check that the chained assignment above changed nothing
print("rows with fare == 99 in titanic:", (titanic["fare"] == 99).sum())    # 0 -> titanic is untouched

# the safe pattern, shown on a copy so that titanic itself is not altered
demo = titanic.copy()
demo.loc[demo["age"] > 60, "fare"] = 99
print("rows with fare == 99 in demo   :", (demo["fare"] == 99).sum())       # the .loc version works
````

### Imputation: what you should think about before filling

- **Median fill shrinks the spread.** Every filled value equals the same number, so the standard deviation drops (see below). This can bias later analyses.
- **Do not impute everything.** Filling `deck` (about 77% missing) would mostly invent data.

- **Leakage.** In machine learning, compute the fill value (median) on the *training* set only and reuse it on the test set; otherwise information leaks from the test data.
- **Better than a global median:** a median per group, e.g. per class and sex. We meet this with `transform()` in Section 9.

## 9. Grouping Data in pandas

Grouping is one of pandas’ most powerful operations. It allows us to **split** a dataset into subsets, **apply** computations independently to each group, and then **combine** the results.  This workflow is known as the **Split → Apply → Combine** pattern.

### Why use grouping?

Grouping helps answer questions such as:
- What is the average fare per passenger class?
- How does the survival rate differ between men and women?
- How many passengers travelled in each combination of class and port?

pandas makes this process easy using the `.groupby()` method.


### How grouping works

When you call:
```python
df.groupby("column_name")
```

pandas creates and returns a special `GroupBy` object — it doesn’t compute anything yet.
You can then perform operations (like `size()`, `mean()`, or `agg()`) on each subgroup independently.

### Common GroupBy Operations

| Operation | Description |
|------------|-------------|
| `df.groupby("col")` | Group by a single column |
| `df.groupby(["col1", "col2"])` | Group by multiple columns |
| `grouped.size()` | Count number of rows in each group |
| `grouped["col"].mean()` | Compute the mean of a column for each group |

````python
# Group by passenger class
grouped = titanic.groupby("class")

# Count passengers per class
print("Passenger count per class:")
print(grouped.size())

# Alternatively, count using a specific column
titanic.groupby("class")["survived"].count()
````

**`size()` vs `count()`.** `size()` counts the **rows** in each group; `count()` counts the **non-missing values** of a column. They are equal for `survived` (no missing values), which hides the difference. For `age`, which has missing values, they differ.

````python
display(pd.DataFrame({
    "size (rows)": grouped.size(),
    "count of survived": grouped["survived"].count(),
    "count of age": grouped["age"].count(),     # fewer: missing ages are not counted
}))
````

### `transform()` and `filter()`: keeping the original shape

`agg()` **reduces** each group to one row. Two other group operations keep the rows:

````python
# group-wise imputation - fill missing ages with the median age of the passenger's own class AND sex
median_age = titanic.groupby(["pclass", "sex"], observed=True)["age"].transform("median")
titanic["age_filled_by_group"] = titanic["age"].fillna(median_age)

display(titanic[["pclass", "sex", "age", "age_filled_by_group"]].head(8))
print("missing before:", titanic["age"].isna().sum(), "| after:", titanic["age_filled_by_group"].isna().sum())

# the group medians that were used (compare with the single global median of Section 7)
display(titanic.groupby(["pclass", "sex"], observed=True)["age"].median().unstack())

# filter: keep only the embarkation towns with more than 100 passengers
big_ports = titanic.groupby("embark_town").filter(lambda g: len(g) > 100)
print(big_ports["embark_town"].value_counts())
````

Aggregation helps you move from raw data to meaningful insights by summarising large tables into compact, interpretable summaries.


An **aggregation function** reduces multiple values in each group down to a single representative value — for example:

This gives you flexibility to compute domain-specific metrics — such as interquartile range or ratios.

````python
# Example 4: Custom aggregation using lambda
custom_agg = (
    titanic
    .groupby("class", observed=True)
    .agg(
        age_iqr=("age", lambda x: x.quantile(0.75) - x.quantile(0.25)),
        fare_range=("fare", lambda x: x.max() - x.min())
    )
    .reset_index()
)

display(custom_agg)
````

### Summary — Grouping vs Aggregation

| Concept | Purpose | Example |
|----------|----------|----------|
| **Grouping** | Splits the data into subsets based on one or more keys | `df.groupby(['class', 'sex'])` |
| **Aggregation** | Summarises each group using statistical functions | `.agg({'age':'mean', 'fare':'sum'})` |

**Grouping** defines *how* the data is divided.  
**Aggregation** defines *what* is computed within each group.

Together, they form the backbone of exploratory data analysis in pandas.

### Pivot Tables

Pivot tables are one of pandas’ most powerful summarisation tools.  
They allow you to **rearrange, aggregate, and summarise** data in a format similar to Excel pivot tables — condensing large datasets into compact summaries that are easy to interpret.


A **pivot table** summarises data by combining four main elements:

| Element | Purpose | Example |
|----------|----------|----------|
| **Index** | Defines the rows of the summary | Passenger `class` |
| **Columns** | Defines the columns of the summary | Passenger `sex` |
| **Values** | What you want to summarise | Mean of `fare` or `survived` |
| **Aggregation Function (aggfunc)** | How values are summarised | `mean`, `count`, `sum`, etc. |

The result is a **two-dimensional table** where each cell represents a statistic for a specific combination of index and column values.

#### **Why use pivot tables?**
-	They provide a compact summary of large datasets.
-	They help uncover relationships between categorical variables.
-	They make quick cross-tab analysis easy — without needing multiple groupby() operations.



For example,
```python
pd.pivot_table(
    data=df,
    values="target_column",
    index="row_variable",
    columns="column_variable",
    aggfunc="mean",
    margins=True
)
```

Here,

| Parameter | Description |
|------------|--------------|
| **`data`** | Your `df` DataFrame containing the data to summarise |
| **`values`** | The numeric column to summarise (e.g., `fare`, `age`, `survived`) |
| **`index`** | Variable to group rows (e.g., `class`, `sex`) |
| **`columns`** | Variable to group columns (e.g., `embarked`, `sex`) |
| **`aggfunc`** | Aggregation function (default = `'mean'`; can also be `'sum'`, `'count'`, etc.) |
| **`margins=True`** | Adds “All” totals for both rows and columns — equivalent to **Grand Totals** in Excel |

---


#### Example — Average Survival Rate by Passenger Class and Sex

Let’s compute the **average survival rate** for each combination of passenger `class` and `sex` using a pivot table. The result will show:
- Rows (`index`) → Passenger class (`First`, `Second`, `Third`)  
- Columns (`columns`) → Sex (`male`, `female`)  
- Values (`values`) → Mean survival rate  
- Grand totals (`margins=True`) → Overall averages across all groups

````python
# Create pivot table
pivot = pd.pivot_table(
    titanic,
    values="survived",
    index="class",
    columns="sex",
    aggfunc="mean",
    margins=True,
    observed=True
)

display(pivot)
````

#### Interpretation

- The **`female`** column has higher survival rates in all classes — reflecting the “women and children first” policy.  
- **First-class** passengers had the highest overall survival rates, followed by **Second** and **Third** class.  
- The **“All”** row and column show overall averages for quick comparison.  

**A note of caution on interpretation.** The table shows *associations* only. The "women and children first" policy is a plausible explanation for the sex difference, but this table does not even contain age, and class, sex and age are tangled together (for example, more men travelled in Third class and in the crew-adjacent cheap cabins). Be careful with causal language, and prefer "is associated with" over "because of".

````python
# Create pivot table for average age by passenger class and sex
pivot_age = pd.pivot_table(
    titanic,
    values="age",
    index="class",
    columns="sex",
    aggfunc="mean",
    margins=True,
    observed=True
)

display(pivot_age)
````

#### Interpretation
- The average age increases **substantially** from Third to First class (about 25 to 38 years) — suggesting older passengers tended to occupy higher classes.
- The male passengers are generally older than female passengers across all classes.
- The “All” totals provide an overall view of the mean age across the entire ship, useful for comparing demographic patterns.

## 11. Reshaping: Wide ↔ Long

Real-world analysis often requires switching between two table shapes:

- **Wide format** (good for *reporting*):  
  Each category becomes a separate **column**.  
  Example: columns `male`, `female` holding survival rates per class.


| class | male_survival | female_survival |
|--------|----------------|----------------|
| First  | 0.37           | 0.97           |
| Second | 0.16           | 0.92           |
| Third  | 0.14           | 0.50           |


- **Long (tidy) format** (good for *modelling/plotting*):  
  One observation per row; categories stored as **values** in a variable column.  
  Example: a single `sex` column with rows `male`/`female`, and a `survival_rate` column.

| class | sex    | survival_rate |
|--------|--------|----------------|
| First  | male   | 0.37 |
| First  | female | 0.97 |
| Second | male   | 0.16 |
| Second | female | 0.92 |
| Third  | male   | 0.14 |
| Third  | female | 0.50 |

>A tidy dataset is one where each column is a variable, each row is an observation, and each type of observation is stored in its own table.

### Why reshape?
- **Modelling** libraries and plotting tools (e.g., Seaborn) prefer **long** tidy data.
- **Dashboards/reports** often prefer **wide** cross-tab summaries.

### Tips: Choose shape based on task
- **Exploratory plotting / ML** → long (tidy) with `melt()`  
- **Human-readable summaries** → wide with `pivot()` or `pivot_table()`  

In the examples below, we’ll:
1) Build a **wide** table of survival rates by `class` and `sex`.  
2) Convert it **back to long** using `melt()`.  
3) Show how to **pivot** a long table to wide.  
4) Demonstrate `stack()`/`unstack()` on a grouped summary.

---





### Going Long with `melt()`
`pd.melt()` turns selected **columns into rows** (wide → long).

**Key parameters**
- `id_vars`: columns to keep fixed (identifiers)
- `value_vars`: columns to unpivot (become rows)
- `var_name`: name for the new “variable” column
- `value_name`: name for the new “value” column

Columns listed in `value_vars` will be stacked under each other and labeled by `var_name`.

---


````python
# 1) Create a WIDE table: survival rate by class (rows) and sex (columns)
wide = pd.pivot_table(
    titanic,
    values="survived",
    index="class",
    columns="sex",
    aggfunc="mean",
    observed=True      # `sex` became a category in Section 7; observed=True avoids a FutureWarning on pandas 2
)
display(wide)
````

````python
# 2) Melt WIDE → LONG (tidy)
long = pd.melt(
    wide.reset_index(),           # ensure 'class' is a column (id var)
    id_vars=["class"],            # keep class fixed
    var_name="sex",               # new column that stores former column labels
    value_name="survival_rate"    # new column for the values
)
display(long)
````

### Merging Tables — Inner, Outer, Left, and Right Joins

Tables could be merged in many different ways. Below, we demonstrate how the four main types of joins behave when applied to subsets of the Titanic dataset.



#### Join Types Overview

| Join Type | Description |
|------------|--------------|
| **Inner Join** | Keeps only rows that exist in **both** DataFrames (intersection). |
| **Left Join** | Keeps all rows from the **left** DataFrame and matches from the right when possible. |
| **Right Join** | Keeps all rows from the **right** DataFrame and matches from the left when possible. |
| **Outer Join** | Keeps **all** rows from both DataFrames, filling missing matches with `NaN`. |

---


We’ll use two small DataFrames based on the Titanic dataset — one with basic passenger details, and another with ticket information.


````python
# Create two small tables
passengers = titanic[["survived", "pclass", "sex", "age"]].head(8).reset_index()
passengers.columns = ["passenger_id", "survived", "class", "sex", "age"]

tickets = pd.DataFrame({
    "passenger_id": [0, 1, 2, 3, 8, 9],  # two IDs (8,9) not in passengers
    "fare": [72.5, 15.0, 7.8, 33.0, 10.5, 20.0],
    "embarked": ["S", "C", "S", "S", "Q", "S"]
})

display(passengers)
display(tickets)
````

#### 1. Inner Join

Keeps only rows where `passenger_id` appears in both DataFrames (the intersection).

````python
inner_join = pd.merge(passengers, tickets, on="passenger_id", how="inner")
display(passengers)
display(tickets)
display(inner_join)
````

#### 2. Left Join
Keeps all rows from `passengers` and adds ticket details where available.

Passengers **without** a matching ticket get `NaN` in the ticket columns (`fare`, `embarked`). Ticket rows with no matching passenger (IDs 8 and 9) are **dropped**.

````python
left_join = pd.merge(passengers, tickets, on="passenger_id", how="left")
display(passengers)
display(tickets)
display(left_join)
````

#### 3. Right Join

Keeps all rows from `tickets`, adding passenger details where available.

Unmatched passenger data appear as NaN.


````python
right_join = pd.merge(passengers, tickets, on="passenger_id", how="right")
display(passengers)
display(tickets)
display(right_join)
````

#### 4. Outer Join

Combines **all rows** from both tables.

Where a match doesn’t exist, missing fields are filled with NaN.


````python
outer_join = pd.merge(passengers, tickets, on="passenger_id", how="outer")
display(passengers)
display(tickets)
display(outer_join)
````

#### Interpretation
- Inner Join → shows only passengers who have ticket records.
- Left Join → keeps all passengers, with missing tickets shown as NaN.
- Right Join → keeps all tickets, even those without matching passengers.
- Outer Join → combines everything — full coverage of both tables.

Merging is one of the most powerful tools in pandas for integrating datasets, allowing you to recreate relational-style data combinations entirely in Python.

---

### Useful merge options: `left_on/right_on`, `suffixes`, `concat`, `indicator`, `validate`

- If the key columns have **different names**, use `left_on=` and `right_on=`.
- If non-key columns share a name, `suffixes=` labels them.
- `pd.concat` **stacks** tables: `axis=0` appends rows, `axis=1` places them side by side.
- `indicator=True` adds a `_merge` column that says where each row came from (`both`, `left_only`, `right_only`). Filtering on it gives an **anti-join** ("rows in A that are *not* in B").
- **Duplicate keys multiply rows.** If a key appears twice in one table, every match is repeated. `validate="one_to_one"` (or `"one_to_many"`, `"many_to_one"`) turns this silent problem into an error.

````python
# 1) different key names + overlapping column names
tickets_r = tickets.rename(columns={"passenger_id": "pid"})
extra = pd.DataFrame({"passenger_id": [0, 1, 2], "age": [99, 98, 97]})            # `age` also exists in passengers

display(pd.merge(passengers, tickets_r, left_on="passenger_id", right_on="pid").drop(columns="pid"))
display(pd.merge(passengers, extra, on="passenger_id"))                                       # default suffixes _x and _y
display(pd.merge(passengers, extra, on="passenger_id", suffixes=("_passenger", "_extra")))  # readable suffixes

# 2) concat: stack rows (axis=0) or columns (axis=1)
display(pd.concat([passengers.head(2), passengers.tail(2)], ignore_index=True))
````

## 13. Advanced Python Techniques with pandas

These Python patterns integrate smoothly with pandas for cleaner, faster, and more expressive code.  
They help you write concise data-manipulation logic without losing readability.  
Always prefer **vectorised** (built-in) operations for performance, but these patterns fill important gaps when custom logic is needed.

### a) List Comprehensions

List comprehensions offer a concise way to create or transform lists in a single readable line.  
They are useful when you need simple element-wise transformations — though in pandas, try to prefer vectorised methods (arithmetic operations, `.str` methods, `np.where`). Note that `.apply()` is **not** vectorised: it still calls a Python function once per element, only hidden in a method.

**Pattern:**  
```python
[new_value for item in iterable if condition]
```

Below we use a list comprehension to flag whether each passenger’s age is above or below the average.

````python
titanic = sns.load_dataset("titanic")

avg_age = titanic["age"].mean()
titanic["age_group"] = ["above_avg" if age > avg_age else "below_avg" for age in titanic["age"]]

display(titanic[["age", "age_group"]].head())
````

**A trap in the example above.** For passengers with a missing age the comparison `age > avg_age` is `False`, so they are silently labelled `below_avg`. A correct label has to treat missing ages explicitly. `np.where(condition, value_if_true, value_if_false)` is the vectorised version of "if / else"; nest it (or use `np.select`, below) for more cases.

````python
print(titanic.loc[titanic["age"].isna(), ["age", "age_group"]].head(3))     # missing ages labelled 'below_avg'!

titanic["age_group_fixed"] = np.where(
    titanic["age"].isna(), "unknown",                                       # first: handle the missing values
    np.where(titanic["age"] > avg_age, "above_avg", "below_avg"))           # then the real comparison
print(titanic["age_group_fixed"].value_counts())
````

### b) `lambda` and `.apply()`

Use `lambda` to define small, anonymous (inline) functions.  
In pandas, combine them with `.apply()` to apply a transformation across a **column** or **row** when vectorisation isn’t straightforward.

Below we compute the length of the text in the `who` column using a `lambda` function applied element-wise (the seaborn Titanic data has no passenger names, so `who` is only an illustration). `Series.apply` calls the function **once per element**; to apply a function to each **row** you would use `DataFrame.apply(..., axis=1)`.

````python
titanic["name_length"] = titanic["who"].apply(lambda x: len(str(x)))
display(titanic[["who", "name_length"]].head())
````

**Compare `apply` with the vectorised alternative, and see a more useful lambda.** The `.str` accessor does the same thing without a Python function. `apply` is the right tool only when no built-in method exists, for example a rule with several branches. `apply(axis=1)` works **row-wise**: the function receives a whole row (use it sparingly, it is the slowest option).

````python
# 1) the same result without apply
print("same as .str.len():", (titanic["who"].apply(lambda x: len(str(x))) == titanic["who"].str.len()).all())

# 2) a more useful lambda: a three-way rule on fare (element-wise)
titanic["fare_band"] = titanic["fare"].apply(lambda f: "cheap" if f < 15 else ("mid" if f < 50 else "expensive"))
print(titanic["fare_band"].value_counts())

# 3) row-wise apply: the function gets a whole row (a Series indexed by column name)
titanic["label"] = titanic.apply(lambda row: f"{row['sex']}-{row['class']}", axis=1)
print(titanic["label"].head(3).tolist())
````

### c) NumPy Vectorisation

Vectorisation uses NumPy and pandas built-in operations that process entire arrays at once, avoiding explicit Python loops.  
It’s both faster and more readable.

Here we compute the **age difference from the average** for all passengers in one vectorised step.

````python
titanic["age_diff"] = titanic["age"] - titanic["age"].mean()
display(titanic[["age", "age_diff"]].head())
````

**NumPy functions work directly on columns.** `np.where` handles two cases; `np.select` handles **several conditions** in order (the first condition that is `True` wins) with a `default` for everything else. Below we also measure the speed difference between a Python loop, `apply` and a vectorised expression on one million values. (Your exact timings will differ; the *ratio* is what matters.)

````python
# np.select - several conditions, first match wins
conditions = [titanic["age"].isna(), titanic["age"] < 16, titanic["age"] < 60]
choices = ["unknown", "child", "adult"]
titanic["life_stage"] = np.select(conditions, choices, default="senior")
print(titanic["life_stage"].value_counts())
````

````python
# why vectorise? Same computation (2x + 1) on 1,000,000 numbers three ways
big = pd.DataFrame({"x": np.random.default_rng(0).normal(size=1_000_000)})

print("Python list comprehension:")
%timeit -n 1 -r 3 [v * 2 + 1 for v in big["x"]]
print("Series.apply with a lambda:")
%timeit -n 1 -r 3 big["x"].apply(lambda v: v * 2 + 1)
print("Vectorised arithmetic:")
%timeit -n 5 -r 3 big["x"] * 2 + 1
````

### d) Method Chaining & `.pipe()`

Method chaining lets you apply multiple pandas transformations step-by-step without creating intermediate variables.  This functional style is more readable and easier to debug.  

Use `.pipe()` to inject custom functions into the chain.

Below we demonstrate filtering, grouping, and transforming in a clean chained expression.

````python
# Two helper functions that we will plug into a chain with .pipe()

def keep_adults(df, min_age=18):
    """Drop rows with unknown age and keep passengers older than min_age."""
    return df.dropna(subset=["age"]).query("age > @min_age")

def survival_by(df, key):
    """Mean survival per group, returned as a one-column DataFrame."""
    return df.groupby(key, observed=True)["survived"].mean().rename("avg_survival").to_frame()

# Example of method chaining with .pipe(): each step takes the DataFrame from the previous step
result = (
    titanic
    .pipe(keep_adults, min_age=18)     # pipe lets you inject your own function (with extra arguments) into the chain
    .pipe(survival_by, "class")
)

display(result)
````

### Anti-patterns to avoid

| Anti-pattern | Why it is a problem | Prefer |
|---|---|---|
| `df.iterrows()` or a `for` loop over rows | very slow (see the timing above) | vectorised column operations |
| `df.apply(func, axis=1)` for simple arithmetic or string joins | one Python call per row | arithmetic, `.str`, `np.where`, `np.select` |
| `inplace=True` | no speed benefit, breaks method chaining, and is being phased out | `df = df.method(...)` |
| `df[mask]["col"] = value` (chained indexing) | never updates `df` (always copy semantics in pandas 3) | `df.loc[mask, "col"] = value` |
| growing a DataFrame inside a loop with `concat` | copies the whole table at every step | collect rows in a list, build the DataFrame **once** |

### e) Working with dates and times

Titanic has no dates, but nearly every real dataset does. A column of dates should have the **`datetime64`** dtype (`pd.to_datetime` converts text). Then:

- the **`.dt` accessor** gives parts of the date (`.dt.year`, `.dt.day_name()`, `.dt.month` ...);
- **`resample`** is `groupby` for time: it aggregates a time-indexed Series into regular periods (`"W"` = weeks, `"MS"` = month starts).

````python
# a small synthetic daily sales table
rng = np.random.default_rng(42)
daily = pd.DataFrame({
    "date": pd.date_range("2024-01-01", periods=120, freq="D"),
    "sales": rng.integers(50, 150, size=120),
})
print(daily.dtypes["date"])

daily["weekday"] = daily["date"].dt.day_name()                  # the .dt accessor
display(daily.groupby("weekday")["sales"].mean().round(1))

monthly = daily.set_index("date")["sales"].resample("MS").sum()   # resample: groupby for time
display(monthly)

# text -> datetime (errors="coerce" turns unparseable entries into NaT, the datetime version of NaN)
print(pd.to_datetime(pd.Series(["2024-03-01", "not a date"]), errors="coerce"))
````

**Interpretation.** `resample("MS")` produced one row per month (January to April) with the total sales; `groupby("weekday")` is not in calendar order (it is alphabetical), so for a report you would `reindex` it with the weekday names in order.

## 14. PySpark for Scalable Data Processing


While **pandas** is ideal for local, in-memory analysis, it struggles with **very large datasets** that don’t fit in RAM.  
For scalability, Python developers often use **[PySpark](https://spark.apache.org/docs/latest/api/python/)** — the Python API for **Apache Spark**, a distributed computing framework.  

PySpark distributes computations across multiple cores or even multiple machines, allowing operations such as filtering, grouping, and aggregation on datasets with **millions or billions of rows**.

Below, we load the Titanic dataset and replicate a pandas-style analysis — this time using PySpark to demonstrate the same logic. (A caveat: Titanic has only 891 rows, so Spark is *slower* than pandas here because of its start-up overhead. We use it to learn the syntax; the benefit only appears with data that does not fit in one machine's memory.)

````python
# PySpark 4.x needs Java 17 or newer, but the default Java on Colab can be older. Check first.
import os, re, subprocess, sys

def java_major_version():
    """Return the major Java version (e.g. 17), or None if Java is not installed."""
    try:
        out = subprocess.run(["java", "-version"], capture_output=True, text=True).stderr
        major = int(re.search(r'version "(\d+)', out).group(1))
        return 8 if major == 1 else major          # old style "1.8.0" means Java 8
    except Exception:
        return None

print("Java major version found:", java_major_version())

if (java_major_version() or 0) < 17 and "google.colab" in sys.modules:
    !apt-get install -qq -y openjdk-17-jdk-headless > /dev/null 2>&1
    os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-17-openjdk-amd64"
    os.environ["PATH"] = os.environ["JAVA_HOME"] + "/bin:" + os.environ["PATH"]
    print("Java after installation:", java_major_version())
````

````python
# Install PySpark (if running in Colab or a local environment)
!pip install -q pyspark

from pyspark.sql import SparkSession
from pyspark.sql import functions as F


# 1 Create a Spark session
spark = SparkSession.builder.appName("Titanic_Scalable").getOrCreate()

# 2 Load Titanic dataset via seaborn (small demo version)
titanic_pd = sns.load_dataset("titanic")

# 3 Convert pandas DataFrame to Spark DataFrame
titanic_spark = spark.createDataFrame(titanic_pd)

# 4 Preview data
titanic_spark.select("pclass", "sex", "age", "survived").show(5)

# 5 Example: Compute survival rate by passenger class and sex
survival_summary = (
    titanic_spark
    .groupBy("pclass", "sex")
    .agg(F.mean("survived").alias("avg_survival_rate"))
    .orderBy("pclass", "sex")
)

survival_summary.show()

# 6 Stop Spark session when done
spark.stop()
````

### Two things the demo above hides

1. **`NaN` is not `null`.** pandas stores a missing number as `NaN`. When you convert a pandas DataFrame, Spark keeps `NaN` as a real floating-point value, **not** as a missing value (`null`). Spark's `mean` then returns `NaN` for any group containing a `NaN`, whereas null values would be ignored (as pandas does).
2. **`createDataFrame(pandas_df)` is not the scalable pattern.** It first builds the whole table in the driver's memory, exactly what Spark is meant to avoid. In practice you let Spark **read the files itself**, ideally Parquet, with `spark.read.csv(...)` or `spark.read.parquet(...)`; Spark then reads the file in parallel partitions.

````python
from pyspark.sql import SparkSession, functions as F
spark = SparkSession.builder.appName("Titanic_Scalable").getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

# (a) from pandas: missing ages are NaN, not null
sdf_from_pandas = spark.createDataFrame(titanic_pd)
print("null ages:", sdf_from_pandas.filter(F.col("age").isNull()).count(),
      "| NaN ages:", sdf_from_pandas.filter(F.isnan("age")).count())
sdf_from_pandas.groupBy("pclass").agg(F.mean("age").alias("mean_age")).orderBy("pclass").show()   # NaN poisons the mean

# (b) the scalable pattern: let Spark read the CSV we wrote in Section 3 -> missing ages become null
sdf = spark.read.csv("titanic.csv", header=True, inferSchema=True)
print("null ages when read from CSV:", sdf.filter(F.col("age").isNull()).count())
sdf.groupBy("pclass").agg(F.round(F.mean("age"), 1).alias("mean_age")).orderBy("pclass").show()      # nulls are ignored

spark.stop()
````

#### Interpretation

- We first created a **SparkSession**, which is the entry point for all PySpark operations.  
- The pandas Titanic dataset was converted to a Spark DataFrame. For real data, read the files with Spark directly instead (see above).
- Using `.groupBy()` and `.agg()`, we computed the **average survival rate** by passenger class and sex — exactly as we would with pandas, but executed in parallel.
- The output mirrors the earlier pivot example in pandas, yet PySpark can handle data **far beyond memory limits**.
- In a real-world setting, this same code could run on a cluster processing **gigabytes or terabytes** of data stored in **Parquet**, **CSV**, or **Delta Lake** formats.

### Beyond Spark: other options when pandas is too slow or too big

| Tool | Idea | When to consider it |
|---|---|---|
| **pandas API on Spark** (`import pyspark.pandas as ps`) | pandas-like syntax, executed by Spark | you know pandas and need a cluster |
| **Polars** | fast, multi-threaded DataFrame library written in Rust, lazy evaluation | data that fits on one machine but pandas is slow |
| **DuckDB** | in-process SQL engine; can query CSV/Parquet files and pandas DataFrames directly | analytical queries on files of many GB on a laptop |
| **Dask** | parallel pandas-like DataFrames across cores or machines | scaling a pandas workflow with few changes |

A sensible order is: *pandas first; if it is too slow or too big on one machine try Polars or DuckDB; use Spark when you genuinely need a cluster.*
