## 2. Variables and data types

A **variable** gives a name to a value. The `=` symbol assigns a value.

Common built-in types include:

- `int`: whole numbers, such as `42`;
- `float`: numbers with a decimal point, such as `3.14`;
- `str`: text, such as `"hello"`;
- `bool`: either `True` or `False`;
- `NoneType`: the special value `None`, often meaning “no value yet”.

Use `type()` to inspect a value's type.

```python
student_name = "Sam"
modules_this_term = 4
hours_of_python = 1.5
has_used_git = False
middle_name = None

print(student_name, type(student_name))
print(modules_this_term, type(modules_this_term))
print(hours_of_python, type(hours_of_python))
print(has_used_git, type(has_used_git))
print(middle_name, type(middle_name))
```

Variable names are case-sensitive and should usually be descriptive. By convention, Python variables use `snake_case`.

```python
module_name = "Programming for Data Science"
Module_Name = "This is a different variable"

print(module_name)
print(Module_Name)
```

## 3. Numbers and operators

Python supports normal arithmetic. The most useful operators are:

- `+`, `-`, `*`, `/` for arithmetic;
- `//` for floor division;
- `%` for remainder;
- `**` for powers.

Parentheses make the intended order of operations clear.

```python
a = 17
b = 5

print("addition:", a + b)
print("subtraction:", a - b)
print("multiplication:", a * b)
print("division:", a / b)
print("floor division:", a // b)
print("remainder:", a % b)
print("power:", b ** 2)
```

Python can convert between compatible types using functions such as `int()`, `float()` and `str()`.

```python
whole_number = int(3.9)
decimal_number = float(7)
number_as_text = str(123)

print(whole_number, type(whole_number))
print(decimal_number, type(decimal_number))
print(number_as_text, type(number_as_text))
```

## 4. Strings

Strings are sequences of characters. You can index them, slice them, combine them, and call useful string methods.

```python
course = "Data Science"

print(course[0])       # first character
print(course[-1])      # last character
print(course[0:4])     # characters at positions 0,1,2,3
print(len(course))
print(course.lower())
print(course.upper())
print(course.replace("AI", "Artificial Intelligence"))
```

An **f-string** lets you insert values into text. Put an `f` before the opening quote and write expressions inside `{}`.

```python
name = "Alex"
score = 87.456

print(f"{name} scored {score} marks.")
print(f"{name} scored {score:.1f} marks to one decimal place.")
```

## 7. Comparisons and `if` statements

Comparisons produce Boolean values (`True` or `False`).

Common comparison operators are `==`, `!=`, `<`, `<=`, `>` and `>=`.

An `if` statement runs code only when a condition is true. **Indentation is part of Python's syntax.**

```python
mark = 72

print(mark >= 50)
print(mark == 72)
print(mark != 0)

if mark >= 70:
    print("Distinction-level mark")
elif mark >= 60:
    print("Merit-level mark")
elif mark >= 50:
    print("Pass-level mark")
else:
    print("Below 50")
```

Conditions can be combined using `and`, `or` and `not`.

```python
attendance = 90
submitted_project = True

if attendance >= 80 and submitted_project:
    print("Both example conditions are satisfied.")
```

## 9. `for` loops

A `for` loop repeats a block of code once for each item in a sequence.

```python
languages = ["Python", "R", "SQL"]

for language in languages:
    print(f"I want to practise {language}.")
```

`range()` is useful when you need a sequence of integers.

```python
for i in range(5):
    print(i)

print("----")

for i in range(1, 6):
    print(i)
```

`enumerate()` gives you both a position and a value while looping.

```python
modules = ["Programming", "Statistics", "Machine Learning"]

for position, module in enumerate(modules, start=1):
    print(position, module)
```

## 10. `while` loops

A `while` loop repeats as long as its condition remains true.

Be careful: if the condition can never become false, you create an **infinite loop**.

```python
count = 3

while count > 0:
    print(count)
    count -= 1

print("Go!")
```

## 11. Functions

A **function** packages reusable behaviour. Functions can accept inputs (parameters) and return outputs.

```python
def greet(name):
    return f"Hello, {name}!"

message = greet("Morgan")
print(message)
```

```python
def mean(values):
    return sum(values) / len(values)

print(mean([10, 20, 30]))
```

Default parameter values make an argument optional.

```python
def describe_mark(mark, pass_mark=50):
    if mark >= pass_mark:
        return f"{mark} is a pass"
    return f"{mark} is below the pass mark"

print(describe_mark(67))
print(describe_mark(67, pass_mark=70))
```

## 12. Reading and writing files

For simple file work, `pathlib.Path` provides a clear cross-platform interface.

The cell below creates a small text file in a folder called `python_intro_files`.

```python
from pathlib import Path

work_dir = Path("python_intro_files")
work_dir.mkdir(exist_ok=True)

notes_path = work_dir / "notes.txt"

notes_path.write_text(
    "Python is readable.\n"
    "Practice matters.\n"
    "Errors are information.\n",
    encoding="utf-8",
)

print("Wrote:", notes_path)
```

Read the whole file with `read_text()`.

```python
contents = notes_path.read_text(encoding="utf-8")
print(contents)
```

For more control, use `open(...)` with a context manager. The file is automatically closed when the indented block ends.

```python
with open(notes_path, "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
```

Writing with mode `"w"` replaces the file. Mode `"a"` appends to the end.

```python
results_path = work_dir / "results.txt"

with open(results_path, "w", encoding="utf-8") as file:
    file.write("student_001,72\n")
    file.write("student_002,65\n")

with open(results_path, "a", encoding="utf-8") as file:
    file.write("student_003,81\n")

print(results_path.read_text(encoding="utf-8"))
```
